```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg", 0)
```

```python
plt.imshow(img, cmap="gray")
plt.title("Grayscale Img")
plt.axis('off')
plt.show()
```

```python
print(f"Shape: {img.shape}")
print(f"Pixel at(0,0): {img[101,101]}")
```

Convert to Binary

```python
_, binary = cv2.threshold(img, 100, 255, cv2.THRESH_BINARY)
plt.imshow(binary, cmap="gray")
plt.show()
```

Negative Tranformation

```python
negative = 255 - img
```

```python
plt.imshow(negative)
plt.show()
```

```python
bright = cv2.add(img, 50)
dark = cv2.subtract(img, 50)
```

```python

plt.imshow(bright)
plt.show()
```

```python

plt.imshow(dark)
plt.show()
```
