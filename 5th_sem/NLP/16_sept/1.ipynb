```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet
from nltk import pos_tag, ne_chunk
```

```python
#download required NLTK Resources

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('maxent_ne_chunker_tab')
```

```python
#part 1 Tokenization and pos tagging

text = "Soham is making coffee"

tokens = word_tokenize(text)

#assign pos tags

tagged_words = pos_tag(tokens)
```

```python
### PART 2) WORD-LEVEL SEMANTIC ANALYSIS

print("\nWord-LEVEL SEMANTIC ANALYSIS")
print("-" * 50)

# Analyze nouns and verbs

for word , tag in tagged_words:

    # Check for noun or verb
    if tag.startswith('NN') or tag.startswith('VB'):

        # Find meaning using Wordnet
        synsets = wordnet.synsets(word)

        if synsets:

            # select the first synset
            synset = synsets[0]

            print("\nWord : ",word)

            # Dsiplay meaning
            print("Meaning : ",synset.definition())

            # Find synonyms
            synonyms = []

            for lemma in synset.lemmas():
                synonyms.append(lemma.name())

            print("Syonyms : ",list(set(synonyms)))

            # Find  hypernymns
            hypernyms = synset.hypernyms()

            if hypernyms:
                print("Hypernyms : ",[h.name().split('.')[0] for h in hypernyms])

            else:
                print("Hypernyms : None")
```

```python
print("\n Word-LEVEL SEMANTIC ANALYSIS")
print("-" * 50)

#analyze nouns in tagged_words

for word, tag in tagged_words:
    #check for noun or verb
    if tag.startswith('NN') or tag.startswith('VB'):

        #Find meanings using Wordnet

        synsets = wordnet.synsets(word)

        if synsets:
            
            #Select First synset
            synset = synsets[0]

            print("\n Word: ", word)

            #Display Warning

            print("Menaing: ", synset.definition())

            #find synonyms

            synonyms = []

            for lemma in synset.lemmas():
                synonyms.append(lemma.name())
            
            print("Synonyms :", list(set(synonyms)))
            #find hypernyms

            hypernyms = synset.hypernyms()

            if hypernyms:
                print("Hypernyms: ", [h.name().split('.')[0] for h in hypernyms])
            
            else:
                print("f off")
```

```python
# Named Entity Recognition

print("\n Named Entity Recognition")
print("-" * 50)

#Identify named entities

named_entities = ne_chunk(tagged_words)

print(named_entities)
```
