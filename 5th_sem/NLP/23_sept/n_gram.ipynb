```python
import nltk
nltk.download('brown')
from nltk.corpus import brown
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

Loading Corpus

```python
corpus = brown.words()

#Case folding and getting vocabulary

lower_case_corpus = [w.lower() for w in corpus]

vocab = set(lower_case_corpus)


print("CORPUS EXAMPLE: ", str(lower_case_corpus[:30]), "\n")
print("VOCAB EXAMPLE: ", str(list(vocab)[:10]))


print("\n Total words in corpus: ", len(lower_case_corpus))
print("Vocab of the corpus: ", len(vocab))
```

Calculate Bigram and Trigram Counts

```python
bigram_counts = {}
trigram_counts ={}

#These Dictionaries will store how many times each bigram occurs and trigram occurs

#Sliding through the corpus to get bigram and trigram counts

for i in range(len(lower_case_corpus) - 2):
    

    #Getting bigram and trigram at each slide

    bigram = (lower_case_corpus[i], lower_case_corpus[i+1])

    trigram = (
        lower_case_corpus[i],
        lower_case_corpus[i+1],
        lower_case_corpus[i+2]
    )

#Keeping track of the bigram counts 


    if bigram in bigram_counts.keys():
        bigram_counts[bigram] += 1
    
    else:
        bigram_counts[bigram] = 1
    

    #Keeping track of hte trigram counts

    if trigram in trigram_counts.keys():
        trigram_counts[trigram] += 1
    
    else:
        trigram_counts[trigram] = 1

print("\n Example, count for bigram('the', 'king') is: ", bigram_counts[('the','king')])
```
