```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = np.random.randint(0,255,(255,200), dtype = np.uint8)
```

```python
print("Pixel Grid [10 x 10]: ")
print(img)
plt.imshow(img, cmap='gray')
plt.show()
```

```python
try:
    img = Image.open("image.jpeg")

    if img.mode != 'L':
        img = img.convert('L')
    
    img_array = np.array(img)

    print("Image Loaded: {img}")
    print(f"Image shape: {img_array.shape}")
    print(f"Pixel Values range: {img_array.min()} to {img_array.max()}")
    print("\n First 10 x 10 pixels (top-left corner): ")
    print(img_array[:10, :10])

    #display the image

    plt.figure(figsize=(8,6))
    plt.imshow(img_array)
    plt.title(f"Loaded Image")
    plt.axis('off')
    plt.show()

except FileNotFoundError:
    print("Error")
```

Increasing Brightness


```python
bright = np.clip(img_array + 50, 0, 255).astype(np.uint8)
```

```python
#display

plt.figure(figsize=(8,6))
plt.imshow(bright)
plt.title("Brigtened Image")
plt.axis('off')
plt.show()
```
