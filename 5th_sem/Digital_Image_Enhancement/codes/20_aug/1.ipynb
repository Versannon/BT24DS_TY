Contrast Stretching

```python
import cv2
import numpy as np
```

```python
img = cv2.imread('image.jpeg', cv2.IMREAD_GRAYSCALE)
```

```python
#Find min and max pixel value
min_val= np.min(img)
max_val=np.max(img)
```

```python
#avoid div by zero
if max_val - min_val ==0:
    stretched = img.copy()
else:
    #Linear stretch: map min + 0 , max -> 255
    stretched = (img - min_val) * (255.0 / (max_val- min_val))
    stretched = np.clip(stretched, 0, 255).astype(np.uint8)
```

```python
#Show result
print(f"Orignal range: {min_val} - {max_val}")
#Show result
cv2.imshow('Original Image', img)
cv2.imshow('Stretched Image', stretched)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
