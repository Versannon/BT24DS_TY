### Introduction to Machine Learning Datasets and Data Exploration using Scikit-learn

#### 1. Aim
To explore the Iris dataset using Scikit-learn and Pandas, understand feature and target separation, observe dataset statistics and class distributions, and perform Train-Test Split for supervised machine learning.

### Step 1: Import Required Libraries
The required Python libraries are imported for data manipulation and analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2: Load Iris Dataset
The Iris dataset is loaded using the Scikit-learn library.

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

### Step 3: Display Dataset Components
The different components available in the Iris dataset are displayed.

```python
print("Feature Names:")
print(iris.feature_names)
print("\nTarget Classes:")
print(iris.target_names)
```

### Step 4: Create DataFrame
The feature data is converted into a Pandas DataFrame for further analysis.

```python
df = pd.DataFrame(
    iris.data,
    columns = iris.feature_names
)
```

### Step 5: Add Target Variable
The target variable representing flower species is added to the DataFrame.

```python
df["Species"] = iris.target
```

### Step 6: Display Dataset Records
The first few records of the dataset are displayed.

```python
df.head()
```

### Step 7: Dataset Exploration
Basic information and statistical summary of the dataset are obtained.

```python
df.info()
df.describe()
df.shape
```

### Step 8: Check Missing Values
The presence of missing values in the dataset is checked.

```python
df.isnull().sum()
```

### Step 9: Check Class Distribution
The number of samples belonging to each target class is observed.

```python
df["Species"].value_counts()
```

### Step 10: Separate Features and Target
The dataset is divided into independent variables (features) and dependent variable (target).

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

### Step 11: Train-Test Split
The dataset is divided into training and testing subsets using Scikit-learn.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

### Step 12: Display Dataset Sizes
The dimensions of training and testing datasets are displayed.

```python
print("Training Data :", X_train.shape)
print("Testing Data :", X_test.shape)
```

### 8. Dataset Visualization
In this experiment, a scatter plot is used to visualize the relationship between sepal length and sepal width of Iris flowers.

```python
plt.figure(figsize=(6,4))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"]
)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Iris Dataset Visualization")
plt.show()
```
